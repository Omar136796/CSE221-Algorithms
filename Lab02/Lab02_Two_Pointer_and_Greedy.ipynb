**Task 01** (Two Pointer)

O(n^2)

```python
def sum(list1,list2):
    length=list1[0]
    total=list1[1]
    flag=False
    for i in range(length):
        sum=0
        for j in range(i+1,length,1):
            if list2[i]+list2[j]==total:
                flag=True
                return f"{i+1} {j+1}"
    if flag==False:
        return "Impossible"

list1=[int(i) for i in input().split(" ")]
list2=[int(j) for j in input().split(" ")]

print(sum(list1,list2))
```

O(n)

```python
def handleFiles(inp1,out1):
    with open(inp1, "r") as infile:
        list1=infile.readline().split(" ")
        length=int(list1[0])
        sum=int(list1[-1])
        res=[]
        list2=list(map(int, infile.readline().split(" ")))

        r=length-1
        l=0
        flag=False
        while l!=r:
            if list2[r]>=sum:
                r-=1
            if list2[r]<sum:
                if list2[r]+list2[l]==sum:
                    flag=True
                    res.append(l+1)
                    res.append(r+1)
                    break
                else:
                    l+=1
        if flag==False:
            res.append("Impossible")
    with open(out1, "w") as outfile:
        for i in res:
            outfile.write(str(i)+" ")




inputFile="input1.txt"
outputFile="output1.txt"

handleFiles(inputFile,outputFile)
```

**Task 02**

O(nlogn)

```python
def mergeSort(arr,L):
    if L<=1:
        return arr
    else:
        left_arr=arr[:L//2:]
        right_arr=arr[L//2::]

        mergeSort(left_arr,L//2)
        mergeSort(right_arr,L//2)

        i=0
        j=0
        k=0
        while i<len(left_arr) and j<len(right_arr):
            if left_arr[i]<right_arr[j]:
                arr[k]=left_arr[i]
                i+=1
            else:
                arr[k]=right_arr[j]
                j+=1
            k+=1

        while i<len(left_arr):
            arr[k]=left_arr[i]
            i+=1
            k+=1
        while j<len(right_arr):
            arr[k]=right_arr[j]
            j+=1
            k+=1
        return arr

N=int(input())
list1=[int(k) for k in input().split(" ")]
M=int(input())
list2=[int(l) for l in input().split(" ")]
n=N+M
list3=list1+list2
print(mergeSort(list3,n))
```

```python
def two_sort(arr1,n,arr2,m):
    l=m+n
    arr3=[0]*l                 # Important
    i=0
    j=0
    k=0
    while i<n and j<m:
        if arr1[i]<arr2[j]:
            arr3[k]=arr1[i]
            i+=1
        else:
            arr3[k]=arr2[j]
            j+=1
        k+=1
    while i<n:
        arr3[k]=arr1[i]
        i+=1
        k+=1
    while j<m:
        arr3[k]=arr2[j]
        j+=1
        k+=1
    return arr3

N=int(input())
list1=[int(k) for k in input().split(" ")]
M=int(input())
list2=[int(l) for l in input().split(" ")]
print(two_sort(list1,N,list2,M))
```

```python
def interval(R,O):
    finish=0
    for r in R:
        start,end=r[0],r[1]
        if finish<=start:
            finish=end
            O.append(r)
    return O

r1=(0,4)
r2=(3,4)
r3=(1,5)
r4=(9,10)
r5=(6,9)
r6=(2,3)
r7=(1,2)

R=[r1,r2,r3,r4,r5,r6,r7]
R.sort(key=lambda x: x[1])
O=[]
print(interval(R,O))
```

```python
def MergeSort(arr):
    L=len(arr)
    if L<=1:
        return arr
    else:
        left_arr=arr[:L//2:]
        right_arr=arr[L//2::]

        MergeSort(left_arr)
        MergeSort(right_arr)

        i=0
        j=0
        k=0
        while i<len(left_arr) and j<len(right_arr):
            if left_arr[i][1]<right_arr[j][1]:
                arr[k]=left_arr[i]
                i+=1
            else:
                arr[k]=right_arr[j]
                j+=1
            k+=1

        while i<len(left_arr):
            arr[k]=left_arr[i]
            i+=1
            k+=1
        while j<len(right_arr):
            arr[k]=right_arr[j]
            j+=1
            k+=1
        return arr

def HandleFiles(inp4,out4):
    with open(inp4, "r") as infile:
        list1=list(map(int,infile.readline().split(" ")))
        work=list1[0]
        people=list1[1]
        list2=[]
        for i in range(work):
            tup1=tuple(map(int,infile.readline().split(" ")))
            list2.append(tup1)
        list3=MergeSort(list2)
        count=0
        finish=[0]*people
        for m in range(len(list3)):
            start,end=list3[m][0],list3[m][1]
            for n in range(people):
                if finish[n]<=start:
                    finish[n]=end
                    count+=1
                    break
    with open(out4, "w") as outfile:
        outfile.write(str(count))







input4="input4.txt"
output4="output4.txt"
HandleFiles(input4,output4)
```

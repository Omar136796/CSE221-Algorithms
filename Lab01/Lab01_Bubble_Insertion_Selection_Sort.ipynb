```python
def handleFiles(iFile,oFile):
    with open(iFile, "r") as infile:
        T=int(infile.readline())

        res=[]
        for _ in range(T):
            num = int(infile.readline())
            if num%2==0:
                res.append(f"{num} is an Even number.")
            else:
                res.append(f"{num} is an Odd number.")

    with open(oFile, "w") as outfile:
        for i in res:
            outfile.write(i + "\n")


inputFile="input1a.txt"
outputFile="output1a.txt"

handleFiles(inputFile,outputFile)
```

**Task 02** (Optimised Bubble Sort)

```python
def handleFiles(inp,out):
    with open(inp,"r") as infile:
        T=int(infile.readline())
        arr1=list(map(int, infile.readline().split(" ")))
        for i in range(T-1):
            flag=False
            for j in range(T-i-1):
                if arr1[j] > arr1[j+1]:
                    arr1[j], arr1[j+1] = arr1[j+1], arr1[j]
                    flag=True
            if flag==False:
                break

    with open(out,"w") as outfile:
        for m in arr1:
            outfile.write(str(m)+" ")


input2="input2.txt"
output2="output2.txt"
handleFiles(input2,output2)
```

**Task 03** (Selection Sort)

```python
def handleFiles(inp,out):
    with open(inp,"r") as infile:
        T=int(infile.readline())
        id=list(map(int, infile.readline().split(" ")))
        marks=list(map(int, infile.readline().split(" ")))
        for i in range(T-1):
            max=i
            for j in range(i+1,T):
                if marks[j]>marks[max]:
                    max=j
                elif marks[j]==marks[max]:
                    if id[j]<id[max]:
                        max=j
            if max!=i:
                marks[i],marks[max]=marks[max],marks[i]
                id[i],id[max]=id[max],id[i]

    with open(out,"w") as outfile:
        for m in range(T):
            outfile.write(f"ID: {id[m]} Mark: {marks[m]}\n")



input="input3.txt"
output="output3.txt"
handleFiles(input,output)
```

**Task 04** (Selection Sort)

```python
def handleFiles(inp,out):
    with open(inp,"r") as infile:
        n = int(infile.readline())
        Names=[]
        Times=[]
        Places=[]
        for i in range(n):
            temp=infile.readline().split(" ")
            Names.append(temp[0])
            Places.append(temp[4])
            time=temp[6]
            time=time[:len(time)-1:]   ###
            hour,min=time.split(":")
            total=int(hour)*60 + int(min)
            Times.append([time,total])
        for i in range(n):
                min = i
                for j in range(i + 1, n):
                    if Names[j]<Names[min]:
                        min = j
                    elif Names[j]==Names[min]:
                        if Times[j][1]>Times[min][1]:
                            min = j
                if min!=i:
                    Names[i], Names[min] = Names[min], Names[i]
                    Times[i], Times[min] = Times[min], Times[i]
                    Places[i], Places[min] = Places[min], Places[i]
    with open(out,"w") as outfile:
        for i in range(n):
            outfile.write(f"{Names[i]} will departure for {Places[i]} at  {Times[i][0]}\n")

input="input4.txt"
output="output4.txt"
handleFiles(input,output)
```

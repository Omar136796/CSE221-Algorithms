```python
def merge(a, b):
    p=len(a)
    q=len(b)
    res=[0]*(p+q)
    i=0
    j=0
    k=0
    while i<len(a) and j<len(b):
        if a[i]<b[j]:
            res[k]=a[i]
            i+=1
        else:
            res[k]=b[j]
            j+=1
        k+=1
    while i<len(a):
        res[k]=a[i]
        i+=1
        k+=1
    while j<len(b):
        res[k]=b[j]
        j+=1
        k+=1
    return res


def mergeSort(arr):
    if len(arr) <= 1:
        return arr
    else:
        mid = len(arr)//2
        left_arr=arr[:mid:]
        right_arr=arr[mid::]
        a1 = mergeSort(left_arr)
        a2 = mergeSort(right_arr)
        return merge(a1, a2)

def handleFiles(inp,out):
    with open(inp, "r") as infile:
        L=int(infile.readline())
        list1=list(map(int, infile.readline().split(" ")))
        res=mergeSort(list1)
    with open(out, "w") as outfile:
        for i in res:
            outfile.write(str(i)+" ")


input="input1.txt"
output="output1.txt"
handleFiles(input,output)
```

```python
def findMax(arr,l,r):
    if l==r:
        return arr[l]
    if l==r-1:
        return max(arr[l],arr[r])
    mid=(l+r)//2
    left_max=findMax(arr,l,mid)
    right_max=findMax(arr,mid+1,r)
    return max(left_max,right_max)

def handleFiles(inp,out):
    with open(inp, "r") as infile:
        N=int(infile.readline())
        list1=list(map(int, infile.readline().split(" ")))
        max1=findMax(list1,0,N-1)
    with open(out, "w") as outfile:
        outfile.write(str(max1))



inp2="input2.txt"
out2="output2.txt"
handleFiles(inp2,out2)
```

```python
def Merge(left, right):
    merged = []
    count = 0
    i = j = 0
    mid=len(left)

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            count += mid - i
            j += 1

    merged.extend(left[i:])
    merged.extend(right[j:])

    return merged, count

def countInversion(arr):
    if len(arr) <= 1:
        return arr, 0

    mid = len(arr) // 2
    left_arr, count_left = countInversion(arr[:mid])
    right_arr, count_right = countInversion(arr[mid:])
    sorted_arr, count = Merge(left_arr, right_arr)

    total_count = count_left + count_right + count
    return sorted_arr, total_count


def handleFiles(inp,out):
    with open(inp, "r") as infile:
        L=int(infile.readline())
        list1=list(map(int, infile.readline().split(" ")))
        res_arr,countI=countInversion(list1)
    with open(out, "w") as outfile:
        outfile.write(str(countI)+" ")


input="input3.txt"
output="output3.txt"
handleFiles(input,output)
```

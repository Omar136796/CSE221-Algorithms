**Heap / Priority Queue**

```python
import heapq        # (min heap er jonno)

data=[10, 20, 43, 1, 2, 65, 17, 44, 2, 3, 1]

heapq.heapify(data)      # 1 (by default min heap banay)
print(data)

a=heapq.heappop(data)    # 2 (pop korle root pop hoye by default tree restructure hoy)

print(a)
print(data)
print()


heapq.heappush(data,1)   # 3 (heap e new element insert)
print(data)
```

```python
import heapq
def dijkstra(adjL,priority_queue,distance):     #uses bfs
    while len(priority_queue)>0:
        u=heapq.heappop(priority_queue)
        w1,n1=u
        for v in adjL[n1]:
            w2,n2=v
            if w1+w2<=distance[n2]:
                distance[n2]=w1+w2
                heapq.heappush(priority_queue,(w1+w2,n2))


def handleFiles(inp,out):
    with open(inp, "r") as infile:
        v,e=map(int, infile.readline().split(" "))
        adjL={i:[] for i in range(1,v+1)}
        for i in range(e):
            start,end,weight=map(int, infile.readline().split(" "))
            adjL[start].append((weight,end))
        root=int(infile.readline())
        distance=[99]*(v+1)
        distance[root]=0
        priority_queue=[(0,root)]

        dijkstra(adjL,priority_queue,distance)

        distance=distance[1:]


        str1=""
        for i in range(len(distance)):
            if distance[i]==99:
                str1+="-1 "
            else:
                str1+=str(distance[i])+" "

        with open(out, "w") as outfile:
            outfile.write(str1)


inp1="input1.txt"
out1="output1.txt"
handleFiles(inp1,out1)
```

```python
import heapq
def dijkstra(adjL,priority_queue,distance):
    while len(priority_queue)>0:
        u=heapq.heappop(priority_queue)
        w1,n1=u
        for v in adjL[n1]:
            w2,n2=v
            if w1+w2<=distance[n2]:
                distance[n2]=w1+w2
                heapq.heappush(priority_queue,(w1+w2,n2))

def HandleFiles(inp,out):
    with open(inp, "r") as infile:
        v,e=map(int, infile.readline().split(" "))
        adjL={i:[] for i in range(1,v+1)}
        for i in range(e):
            start,end,weight=map(int, infile.readline().split(" "))
            adjL[start].append((weight,end))
        s1,s2=map(int, infile.readline().split(" "))
        distance1=[99]*(v+1)
        distance2=[99]*(v+1)
        distance1[s1]=0
        distance2[s2]=0
        priority_queue1=[(0,s1)]
        priority_queue2=[(0,s2)]

        dijkstra(adjL,priority_queue1,distance1)
        dijkstra(adjL,priority_queue2,distance2)

        distance1=distance1[1:]
        distance2=distance2[1:]


        pair_max=[]


        for i in range(len(distance1)):
            pair_max.append(max(distance1[i],distance2[i]))


        node=1
        time=pair_max[0]
        for i in range(1,len(pair_max)):
            if pair_max[i]<time:
                time=pair_max[i]
                node=i+1

        flag=False
        for i in range(len(pair_max)):
            if pair_max[i]!=99:
                flag=True
                break

    with open(out,"w") as outfile:
        if flag==False:
            outfile.write(str("Impossible"))
        else:
            outfile.write(f"Time {time}\nNode {node}")



inp1="input2.txt"
out1="output2.txt"
HandleFiles(inp1,out1)
```

```python
import heapq
import math
def dijkstra2(adjL,priority_queue,distance):
    while len(priority_queue)>0:
        u=heapq.heappop(priority_queue)
        w1,n1=u
        for v in adjL[n1]:
            w2,n2=v
            if max(w1,w2)<=distance[n2]:                              # change in dijkstra
                distance[n2]=max(w1,w2)                               # change in dijkstra
                heapq.heappush(priority_queue,(max(w1,w2),n2))        # change in dijkstra

def handle_Files(inp,out):
    with open(inp, "r") as infile:
        v,e=map(int, infile.readline().split(" "))
        adjL={i:[] for i in range(1,v+1)}
        for i in range(e):
            start,end,weight=map(int, infile.readline().split(" "))
            adjL[start].append((weight,end))

        distance=[math.inf]*(v+1)
        distance[1]=0
        priority_queue=[(0,1)]

        dijkstra2(adjL,priority_queue,distance)

        distance=distance[1:]
        #print(distance)

        str1=""
        flag=True
        if distance[-1]==math.inf:
            flag=False
            str1="Impossible"
        else:
            str1=str(distance[-1])
    with open(out, "w") as outfile:
        outfile.write(str1)


inp1="input3.txt"
out1="output3.txt"
handle_Files(inp1,out1)
```

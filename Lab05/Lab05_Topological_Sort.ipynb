**Task 01**

Using DFS

```python
counter=1

def topological_sort_dfs(u,visited,start,end,adj):
    global counter
    if visited[u]!=True:
        visited[u]=True
        start[u]=counter
        counter+=1
        for v in adj[u]:
            if visited[v]==True:
                continue
            topological_sort_dfs(v,visited,start,end,adj)
        end[u]=counter
        counter+=1

def handleFiles(inp,out):
    with open(inp,"r") as infile:
        v,e=map(int,infile.readline().split(" "))
        adj={i:[] for i in range(1,v+1)}
        arr1=[]
        for i in range(e):
            m,n=map(int,infile.readline().split(" "))
            adj[m].append(n)
            arr1.append((m,n))
        visited={vis:False for vis in range(1,v+1)}
        start={s:0 for s in range(1,v+1)}
        end={e:0 for e in range(1,v+1)}

        for j in range(1,v+1):
            topological_sort_dfs(j,visited,start,end,adj)

        end2=sorted(end.items(), key=lambda x:x[1])
        end2=end2[::-1]
        res=""

        print(end)
        print(end2)

        for i in end2:
            res+=str(i[0])+" "

        flag=True
        for i in range(e):
            e1,e2=arr1[i][0],arr1[i][1]
            if end[e2]>end[e1]:
                flag=False
                break

    with open(out,"w") as outfile:
        if flag==False:
            outfile.write("Impossible")
        else:
            outfile.write(res)


inp1="input1a.txt"
out1="output1a.txt"
handleFiles(inp1,out1)
```

Using BFS

```python
def bfs(visited,queue,adjL,node_indegree,arr1):
    for v in queue:
        visited[v]=True
        while len(queue)>0:
            u=queue.pop(0)
            arr1.append(u)
            for v2 in adjL[u]:
                if visited[v2]==True:
                    continue
                else:
                    node_indegree[v2]-=1
                    if node_indegree[v2]==0:
                        queue.append(v2)



def HandleFiles(inp,out):
    with open(inp,"r") as infile:
        v,e=map(int,infile.readline().split(" "))
        adjL={i:[] for i in range(1,v+1)}
        visited={vis:False for vis in range(1,v+1)}
        node_indegree={n:0 for n in range(1,v+1)}
        for i in range(e):
            m,n=map(int,infile.readline().split(" "))
            adjL[m].append(n)
            node_indegree[n]+=1
        queue=[]
        arr1=[]
        for k,v3 in node_indegree.items():
            if v3==0:
                queue.append(k)
        bfs(visited,queue,adjL,node_indegree,arr1)

    with open(out,"w") as outfile:
        str1=""
        if len(arr1)==v:
            for i in arr1:
                str1+=str(i)+" "
        else:
            str1="Impossible"
        outfile.write(str1)




input="input1b.txt"
output="output1b.txt"
HandleFiles(input,output)
```

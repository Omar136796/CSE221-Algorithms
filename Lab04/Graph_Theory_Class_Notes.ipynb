**DFS** Complexity (O(v+2e))

```python
visited=[False for i in range(10)]
adj={1:[2,3,4],2:[1,5],3:[1,6],4:[1,7],5:[2],6:[3,8],7:[4],8:[6]}



def DFS(u):
    #seg 1 : Just after entering a node
    visited[u]=True
    print(u)
    for v in adj[u]:
        #seg 2 : Just before entering a neighbour
        if visited[v]==True:
            continue
        DFS(v)
        #seg 3 : Just after exiting a neighbour
    #seg 4 : Just before exiting a node
```

***Class 3,4***

**BFS**

```python
visited=[False for i in range(10)]
adj={1:[2,3,4],2:[1,5,6],3:[1,7],4:[1,8],5:[2],6:[2,9],7:[3],8:[4],9:[6]}

def BFS(s):
    queue=[]
    visited[s]=True
    queue.append(s)

    while len(queue)>0:
        u=queue.pop(0)  # First in first out    # eikhane jodi u=queue.pop() dei taile DFS hoye jabe [but order ta ulta hobe]
        print(u,end=" ")

        for v in adj[u]:              # visiting all the neighbour
            if visited[v]==True:
                continue
            queue.append(v)
            visited[v]=True
```

```python
BFS(1)
```

**BFS Application**

Node Distance from source. (Shortest Distance o ber kora jabe eta diye)

```python
visited=[False for i in range(10)]
adj={1:[2,3,4,8],2:[1,5,6],3:[1,7],4:[1],5:[2],6:[2,9],7:[3,8],8:[1,7],9:[6]}

distance=[999999 for i in range(10)]

def bfs_distance(s):
    queue=[]
    visited[s]=True
    queue.append(s)
    distance[s]=0      # Important : disconnected graph er distance infinity hobe

    while len(queue)>0:
        u=queue.pop(0)  # First in first out
        print(u,end=" ")

        for v in adj[u]:
            if visited[v]==True:
                continue                  # ei line er jonno shortest distance o solve hobe. karon agei visited hoye jabe so r dhukbe na
            queue.append(v)
            visited[v]=True
            distance[v]=distance[u]+1

bfs_distance(1)          # age call diye nite hobe or distance list update hobe na
```

```python
for i in range(1,10):
    print(f"Distance of node {i} = {distance[i]}")
```

Shortest path using BFS

```python
visited=[False for i in range(10)]
adj={1:[2,3],2:[1,4,5],3:[1,4],4:[2,3,6],5:[2,7],6:[4,7],7:[5,6],8:[9],9:[8]}

distance=[999999 for i in range(10)]
parent=[-1 for i in range(10)]


def bfs_shortest_path(s):
    queue=[]
    visited[s]=True
    queue.append(s)
    distance[s]=0

    while len(queue)>0:
        u=queue.pop(0)
        #print(u,end=" ")

        for v in adj[u]:
            if visited[v]==True:
                continue
            queue.append(v)
            visited[v]=True
            distance[v]=distance[u]+1
            parent[v]=u                # parent update hoitese. Source er parent khali update hobe na (-1 e thakbe)

bfs_shortest_path(1)


source=1
destination=6
current=destination
path=[]

while current!=source:
    path.append(current)
    current=parent[current]
path.append(source)
print(path[::-1])
```

Visiting time stamps using DFS

```python
def DFS2(u):
    #seg 1 : Just after entering a node
    global counter
    visited[u]=True
    start[u]=counter
    counter+=1
    #print(u,end=" ")
    for v in adj[u]:
        #seg 2 : Just before entering a neighbour
        if visited[v]==True:
            continue
        DFS2(v)
        #seg 3 : Just after exiting a neighbour
    #seg 4 : Just before exiting a node
    end[u]=counter
    counter+=1

#DFS2(1)
```

```python
str1=""
def dfs(u,vis,adjL):
    global str1
    vis[u]=True
    str1+=str(u)+" "
    for k in adjL[u]:
        if vis[k]==False:
            dfs(k,vis,adjL)
    return str1

def handleFiles(inp,out):
    with open(inp,"r") as infile:
        v,e=map(int,infile.readline().split(" "))
        adjList={i:[] for i in range(1,v+1)}
        visited=[False for i in range(1,v+2)]
        for i in range(e):
            u,v=map(int,infile.readline().split(" "))
            adjList[u].append(v)
            adjList[v].append(u)

    with open(out, "w") as outfile:
        s=dfs(1,visited,adjList)
        outfile.write(str(s))


inp1="input3.txt"
out1="output3.txt"
handleFiles(inp1,out1)
```
